#The Conceptual Overview

###The Goal
A neural network is essentially a mathematical machine that takes in data ($X$), makes a guess ($\hat{Y}$), checks how wrong it is, and then adjusts its internal "knobs" to make a better guess next time.

###The 5 Phases of the Learning Cycle
1. Initialization (Setting the Knobs)Before the network can learn, it needs a starting point. We create Weights ($W$) and Biases ($b$).Weights are initialized with small, random numbers so that different neurons learn different things.Biases can start at zero.
2. Forward Propagation (Making a Guess)The data moves from the input layer to the output layer. For every layer, we do two things:The Linear Step: Multiply inputs by weights and add the bias.$$Z = W \cdot X + b$$The Activation Step: Pass $Z$ through a function (like the Sigmoid function) to squash the number between 0 and 1, adding non-linearity.$$A = \sigma(Z)$$
3. Cost Calculation (Measuring the Error)Once the network outputs its final guess ($A^{[L]}$), we compare it to the actual true label ($Y$). We use a Cost Function (like Binary Crossentropy) to calculate a single number that represents the total error. The goal of the network is to drive this number as close to zero as possible.
4. Backward Propagation (Finding the Blame)This is how the network learns. Using calculus (the Chain Rule), we move backward from the final error to the input. We calculate Gradients (labeled as $dW$ and $db$). These gradients tell us exactly how much each specific weight and bias contributed to the final error, and in which direction we need to change them.
5. Parameter Update (Taking a Step)We use Gradient Descent to update our actual weights and biases. We subtract a small fraction (the Learning Rate, $\alpha$) of the gradient from the current weights to move "downhill" toward a lower error.$$W = W - \alpha \cdot dW$$(This cycle repeats thousands of times until the network is highly accurate).

#Code

```python
import numpy as np

# ==========================================
# 0. SETUP & HELPER FUNCTIONS
# ==========================================
def sigmoid(z):
    """Squashes numbers between 0 and 1"""
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(a):
    """Calculates the slope of the sigmoid function for backprop"""
    return a * (1 - a)

# Dataset: 2 Training Examples (m=2), 2 Features each
X = np.array([[0.0, 1.0],
              [0.0, 1.0]])

# Target labels (We want the network to output 0 for the first, 1 for the second)
Y = np.array([[0.0, 1.0]])

m = X.shape[1] # Number of examples (2)
learning_rate = 0.5

# Initialize Weights and Biases (Using fixed small numbers for testing)
# Layer 1 (Hidden Layer): 2 Neurons
W1 = np.array([[0.2, -0.1],
               [0.5,  0.4]])
b1 = np.array([[0.1],
               [-0.1]])

# Layer 2 (Output Layer): 1 Neuron
W2 = np.array([[0.3, -0.4]])
b2 = np.array([[0.2]])

print("🚀 STARTING NEURAL NETWORK: 1 EPOCH 🚀\n")

# ==========================================
# 1. FORWARD PROPAGATION
# ==========================================
# Layer 1
Z1 = np.dot(W1, X) + b1
A1 = sigmoid(Z1)

# Layer 2
Z2 = np.dot(W2, A1) + b2
A2 = sigmoid(Z2) # This is our final prediction!

print(f"Network's Initial Guesses:\n{A2}\n")

# ==========================================
# 2. COST CALCULATION
# ==========================================
cost = -1/m * np.sum(Y * np.log(A2) + (1 - Y) * np.log(1 - A2))
print(f"📉 Total Cost (Error): {cost:.4f}\n")

# ==========================================
# 3. BACKWARD PROPAGATION
# ==========================================
# Layer 2 Gradients
dA2 = -(np.divide(Y, A2) - np.divide(1 - Y, 1 - A2))
dZ2 = dA2 * sigmoid_derivative(A2)
dW2 = 1/m * np.dot(dZ2, A1.T)
db2 = 1/m * np.sum(dZ2, axis=1, keepdims=True)

# Layer 1 Gradients
dA1 = np.dot(W2.T, dZ2)
dZ1 = dA1 * sigmoid_derivative(A1)
dW1 = 1/m * np.dot(dZ1, X.T)
db1 = 1/m * np.sum(dZ1, axis=1, keepdims=True)

# ==========================================
# 4. UPDATING PARAMETERS
# ==========================================
# Subtracting the (learning rate * gradient) from original weights
W1 = W1 - learning_rate * dW1
b1 = b1 - learning_rate * db1

W2 = W2 - learning_rate * dW2
b2 = b2 - learning_rate * db2

print("🔄 WEIGHTS UPDATED SUCCESSFULLY!")
print(f"New W2 values:\n{W2}")
```
